```python
# Radiation pressure noise
import numpy as np
import matplotlib.pyplot as plt
from pykat import finesse        # import the whole pykat.finesse package
from pykat.commands import *     # import all packages in pykat.commands
```

```python
pi = 3.14159
hbar = 1.05e-34
c = 3e8

lambda_em = 1064e-9
omega_em = 2*pi*c/lambda_em

P = 1e6
mass = 40

frq = np.logspace(0,3,100)

rpn_ana = np.sqrt(hbar*omega_em*P)/(2*mass*c*pi**2*frq**2)

basekat=finesse.kat() # initialising Finesse
basekat.verbose = False
basecode = """
l laser 100 0 n0   # Laser (Power = 1 W, wavelength offset = 0)
s s1 1 n0 nc1    # Space (Length = 1 m)

## The cavity ##
m m1 0.7 0.3 0.1 nc1 nc2  # Mirror  (R = 0.7, T = 0.3, phi = 0)
s sL 4000 nc2 nc3       # Space (Length = 4 km)
m m2 0.8 0.2 0 nc3 nc4  # Mirror  (R = 0.8, T = 0.2, phi = 0)
attr m1 mass 40 # kg
attr m2 mass 40 # kg

## Detectors ##
# Photo diodes measuring DC-power
pd refl nc1        # Reflected 
pd circ nc2        # Circulating
pd tran nc4        # Transmitted

#cav myCav m1 nc2 m2 nc3

## Simulation instructions ##


fsig noise 1
qnoised qn 1 $fs nc2
xaxis noise f log 1 1000  100

yaxis abs                        # Returns magnitude of detector outputs


"""
# Parsing the FINESSE code
basekat.parse(basecode)
out1 = basekat.run()

plt.loglog(frq,rpn_ana)
plt.show()

plt.loglog(out1.x,out1['qn'])
plt.show()
```
